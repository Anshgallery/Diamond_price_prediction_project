````python
# extracted from unique one 
color_cat =['F', 'J', 'G', 'E', 'D', 'H', 'I']
cut_cat =['Premium', 'Very Good', 'Ideal', 'Good', 'Fair']
clarity_cat =['VS2', 'SI2', 'VS1', 'SI1', 'IF', 'VVS2', 'VVS1', 'I1']
````

````python
from sklearn.impute import SimpleImputer # handle missing values 
from sklearn.preprocessing import StandardScaler # handle feature scaling
from sklearn.preprocessing import OrdinalEncoder # handel encoding

## pipelines
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
````

# **Pipeline** :**It performs multiple preprocessing steps one after another**

````python
numerical_pipeline = Pipeline(

    steps=(
           ("imputer",SimpleImputer()), # to handle missing value first 
           ("scaling",StandardScaler()) # then scalling
    )
)
````

````python
categorical_pipeline = Pipeline(

    steps=(
            ("imputer",SimpleImputer(strategy="most_frequent")),
            ("encoding",OrdinalEncoder(categories= [cut_cat,color_cat,clarity_cat] ))
        
             
         )

    )
````

**("name", (pipeline, columns))**

````python
preprocessor =ColumnTransformer(

            [
                ("numerical_pipeline process",numerical_pipeline,numerical ), # APPLY NUMERICAL COLUMN ON NUMERICAL_PIPELINE : (ANYTHING,PIPELINE,COLUMNS )
                ("categorical_pipeline_process",categorical_pipeline,category)
                    
            ]   

)
````

````python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(independent,dependent,test_size=0.30,random_state=42)
````

````python
x_train
````

````python
x_train =preprocessor.fit_transform(x_train) # FIRST LEARN FROM X_TRAIN MEANS COLUMN TRANSFORMER FIRST GO AND FIT IN PIPERINES THEN PERFORM CHANGES THROUGH TRANSFORM
````

# SimpleImputer Cheat Sheet

## 1. Import

```python
from sklearn.impute import SimpleImputer
```

---

## 2. Numerical Columns (Mean)

```python
num_imputer = SimpleImputer(strategy="mean")

df[num_cols] = num_imputer.fit_transform(df[num_cols])
```

---

## 3. Numerical Columns (Median)

```python
num_imputer = SimpleImputer(strategy="median")

df[num_cols] = num_imputer.fit_transform(df[num_cols])
```

---

## 4. Categorical Columns (Most Frequent)

```python
cat_imputer = SimpleImputer(strategy="most_frequent")

df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

---

## 5. Constant Value

### Numerical

```python
num_imputer = SimpleImputer(strategy="constant", fill_value=0)
```

### Categorical

```python
cat_imputer = SimpleImputer(strategy="constant", fill_value="Unknown")
```

---

## 6. Automatically Select Columns

```python
num_cols = df.select_dtypes(include="number").columns

cat_cols = df.select_dtypes(include="object").columns
```

---

## 7. Complete Example

```python
from sklearn.impute import SimpleImputer

num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include="object").columns

num_imputer = SimpleImputer(strategy="median")
cat_imputer = SimpleImputer(strategy="most_frequent")

df[num_cols] = num_imputer.fit_transform(df[num_cols])
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

---

## 8. Machine Learning Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", SimpleImputer(strategy="median"), num_cols),
        ("cat", SimpleImputer(strategy="most_frequent"), cat_cols)
    ]
)
```

---

## Strategies to Remember

| Data Type | Strategy |
|-----------|----------|
| Numeric | `"mean"` |
| Numeric (with outliers) | `"median"` |
| Categorical | `"most_frequent"` |
| Custom Value | `"constant"` + `fill_value=` |

### Memory Trick

- **Mean** → Normally distributed numeric data
- **Median** → Numeric data with outliers ✅ (most common)
- **Most Frequent** → Categorical data ✅
- **Constant** → Replace with your own value (e.g., `0` or `"Unknown"`)

````python
class Employee:

    company = "Google" # class variable

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # Instance Method
    def display(self):
        print(self.name, self.salary)

    # Class Method
    @classmethod
    def change_company(cls, company):
        cls.company = company

    # Static Method
    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10


emp1 = Employee("Ansh", 50000)
emp2 = Employee("Rahul", 60000)

emp1.display()
emp2.display()

print(Employee.company)

Employee.change_company("Microsoft") # access directly by class not by object 

print(Employee.company)

print(Employee.calculate_tax(50000))
````
