## ***What is absolute path and relative path***


# Path

A **path** is the address (location) of a file or folder on your computer.

**Example:**
```python
C:\Users\Ansh\Desktop\Project\data\train.csv
```

---

# Absolute Path

- Complete address of a file.
- Starts from the drive (C:, D:, etc.).
- Works from anywhere.

**Example:**
```python
C:\Users\Ansh\Desktop\Project\data\train.csv
```

---

# Relative Path

- Address of a file relative to the current folder.
- Makes projects portable and easy to share.

**Examples:**
```python
data/train.csv      # File inside data folder
../data/train.csv   # Go one folder up, then into data
./train.csv         # File in the current folder
```

---

# Symbols

- `.` → Current folder
- `..` → One folder up (parent folder)
- `../..` → Two folders up

---

# Interview Point

- **Absolute Path:** Full location of a file.
- **Relative Path:** Location relative to the current working directory. Preferred in ML and Python projects because it works on different computers.

````python
plt.figure(figsize=(30,20))
sns.regplot(x="carat",y="price",data=df) # TO CHECK LINEAR RELATIONSHIP
sns.scatterplot(x="carat",y="price",data=df,hue="clarity")
````

A strong positive linear relationship exists between carat and price across all clarity grades. However, one SI1 diamond of approximately 0.7 carat is priced around 18,000, which is significantly higher than the average price predicted by the regression line (approximately 5,000). This observation lies well above the fitted trend, suggesting that additional factors such as cut, color, depth, table, or other quality attributes may have contributed to its premium price.
The majority of diamonds are concentrated between 1.0 and 1.8 carats, indicating that this is the most common carat range in the dataset.
A relatively high concentration of SI2 diamonds is observed in the 2.0–2.5 carat range compared with several other clarity grades. This suggests greater availability of SI2 diamonds within this carat interval in the dataset.

# **use map to encode**

````python
cut_map ={"Fair":1,"Good":2,"Very Good":3,"Premium":4,"Ideal":5}
clarity_mape={"I1":1,"SI2":2,"SI1":3,"VS2":4,"VS1":5,"VVS2":6,"VVS1":7,"IF":8}
color_map ={"J":1,"I":2,"H":3,"G":4,"F":5,"E":6,"D":7}
````

````python
df["cut"] =df["cut"].map(cut_map)
df["clarity"]=df["clarity"].map(clarity_mape)
df["color"] = df["color"].map(color_map)
````
